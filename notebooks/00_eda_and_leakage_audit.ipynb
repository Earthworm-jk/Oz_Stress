## Leakage checklist

- `ID`는 feature에서 제외합니다.
- categorical 결측값은 train fold에서 fit되는 Pipeline 내부에서 처리합니다.
- `mean_working` 중앙값은 `StressFeatureEngineer.fit()`에서 train fold 기준으로만 계산됩니다.
- validation/test에는 transform만 적용됩니다.
- test는 최종 submission 생성 단계에서만 predict에 들어갑니다.
